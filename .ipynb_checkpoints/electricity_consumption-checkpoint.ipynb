Problem Statement
You are provided with electricity consumption data (in kWh) for a household measured daily over 4 months (120 days).
 The daily consumption values range between 200 kWh (minimum) and 600 kWh (maximum).
Your task is to simulate the dataset using NumPy and perform an in-depth analysis of the household’s electricity consumption.

Tasks (Do all 10)
1. Generate the Dataset
Use numpy.random.randint() to simulate 120 daily consumption values between 200 and 600 kWh.
Store the data in a NumPy array called consumption.

2. Basic Statistics
Calculate the mean, median, minimum, maximum, and standard deviation of daily consumption.

3. Identify Extremes
Find the day number(s) with the highest and lowest consumption.

4. Monthly Splitting
Divide the dataset into 4 months (30 days each).
Store each month’s data in a separate NumPy array.

5. Monthly Analysis
For each month: calculate the total consumption and average daily consumption.

6. Weekly Analysis
Reshape or split the data into weeks of 7 days (ignore the extra 4 days at the end).
Compute the weekly totals and weekly averages.

7. Overall Consumption Trend
Calculate the cumulative consumption across the 120 days.
Display the first 10 values of the cumulative sum.

8. Anomaly Detection
Define anomalies as days where consumption is 2 standard deviations above or below the mean.
Identify the day numbers that are anomalies.




```python
import numpy as np
import pandas as pd
# 1. Generate the Dataset
# Use numpy.random.randint() to simulate 120 daily consumption values between 200 and 600 kWh.
# Store the data in a NumPy array called consumption.

np.random.seed(42)
days = np.arange(1 , 121)
consumption = np.random.randint(200 , 600, size=120)

print("Days:", days)
print("\nConsumption:", consumption)
```

```python
# Store the data in a NumPy array called consumption.
data = pd.DataFrame({
    "Days" : days, 
    "Consumption (kWh)" : consumption

})

data
```

```python
#2. Basic Statistics
# Calculate the mean, median, minimum, maximum, and standard deviation of daily consumption.

Mean = np.mean(consumption)
Median = np.median(consumption)
Minimum = np.min(consumption)
Maximum = np.max(consumption)
Std = np.std(consumption)

print("Mean_Consumption:", round(Mean, 2), "kWh")
print("Median_Cosumption:", round(Median, 2),"kWh")
print("Minimum_Consumption:", round(Minimum, 2),"kWh")
print("Maximum_Consumption:", round(Maximum, 2),"kWh")
print("Std_Consumption:", round(Std, 2),"kWh")

```

```python
#3. Identify Extremes
# Find the day number(s) with the highest and lowest consumption.
            # np.argmax() → returns the index of the largest value. # +1 for human-readable day count 
            # np.argmin() → returns the index of the smallest value. # +1 for human-readable day count
            # np.max() → return the actual highest values.
            # np.min() → return the actual lowest values.

#highest = np.where(consumption == np.max(consumption))[0] + 1
#print('Days with Highest Consumption:', highest)

#lowest = np.where(consumption == np.min(consumption))[0] + 1
#print('Days with Highest Consumption:', lowest)


highest_consumption_day = np.argmax(consumption)
lowest_consumption_day = np.argmin(consumption)

print("Highest Consumption Day Number:", highest_consumption_day + 1)
print("Lowest Consumption Day Number:", lowest_consumption_day + 1)




```

```python
# 4. Monthly Splitting
# Divide the dataset into 4 months (30 days each).


Month = np.array_split(consumption, 4)

print(Month)
```

```python
# Store each month’s data in a separate NumPy array.
month1, month2, month3, month4 = np.array_split(consumption, 4)
print("First_Month:", month1)
print("Second_Month:", month2)
print("Third_Month:", month3)
print("Forth_Month:", month4)
```

```python
# 5. Monthly Analysis
# For each month: calculate the total consumption and average daily consumption.

#print('Monthly Total Consumptions:')
#monthly_total_consump = [print(month.sum()) for month in months] 

#print('\nAverage Consumptions per Month:')
#monthly_avg_consump = [print(month.mean()) for month in months]

print("Monthly Total Consumption\n")

print("First-Month-Total-Consumption:", np.sum(month1),"kWh")
print("Second-Month-Total-Consumption:", np.sum(month2),"kWh")
print("Third-Month-Total-Consumption:", np.sum(month3),"kWh")
print("Forth-Month-Total-Consumption:", np.sum(month4),"kWh")

print("\nAverage Daily Consumption\n")

print("First-Month-Average-Daily-Consumption:", round(np.mean(month1), 2),"kWh")
print("Second-Month-Average-Daily-Consumption:", round(np.mean(month2), 2),"kWh")
print("Third-Month-Average-Daily-Consumption:", round(np.mean(month3), 2),"kWh")
print("Forth-Month-Average-Daily-Consumption:", round(np.mean(month4), 2), "kWh")


```

```python
# 7. Overall Consumption Trend
# Calculate the cumulative consumption across the 120 days.
# Display the first 10 values of the cumulative sum.# Example 120-day consumption dataset

# Calculate cumulative consumption
cumulative_consumption = np.cumsum(consumption)

# Display results
print("Daily Consumption:")
print(consumption)

print("\nCumulative Consumption:")
print(cumulative_consumption)
```

```python
# Display the first 10 values of the cumulative sum

# Display first 10 values
print(cumulative_consumption[:10])
```

```python
#8. Anomaly Detection
#Define anomalies as days where consumption is 2 standard deviations above or below the mean.


# Define anomaly thresholds
upper_bound = Mean + (2 * Std)
lower_bound = Mean - (2 * Std)


print("\nUpper Bound:", round(upper_bound, 2))
print("Lower Bound:", round(lower_bound, 2))

```

```python
# Identify the day numbers that are anomalies.

anomaly_days = np.where(
    (consumption > upper_bound) | (consumption < lower_bound))[0] + 1 

print("\nAnomaly Days:", anomaly_days)

```
